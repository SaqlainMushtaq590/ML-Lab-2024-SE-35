## Task 8 · Inflation FY2024 vs FY2025 (real data, Pakistan Economic Survey 2024-25)

- **(e)** Label each FY2025 month "Low" (< 5), "Moderate" (5–10) or "High" (≥ 10) with `np.where`.

````python
# T8: your code here
avg_inflation = np.mean(cpi, axis=1)
print("T8a - Average Inflation (FY2024, FY2025):", avg_inflation)

fy24_above_25 = cpi_months[cpi[0] > 25]
print("T8b - FY24 months > 25%:", fy24_above_25)

inflation_drop = cpi[0] - cpi[1]
print("T8c - Inflation Drop (FY24 - FY25):", inflation_drop)

biggest_drop_month = cpi_months[np.argmax(inflation_drop)]
print("T8d - Month with biggest drop:", biggest_drop_month)

fy25_labels = np.where(cpi[1] < 5, "Low", np.where(cpi[1] <= 10, "Moderate", "High"))
print("T8e - FY2025 Inflation Labels:", fy25_labels)
````

## Task 9 · Grace marks
Using `marks`:
- **(a)** Add 5 grace marks but cap at 100 (`np.clip`).
- **(b)** Pass % (≥ 50) after grace.
- **(c)** Index of students who passed **only because of** the grace marks.
- **(d)** z-score of each student after grace.

````python
# T9: your code here
marks_with_grace = np.clip(marks + 5, 0, 100)
print("T9a Marks with Grace (capped at 100):", marks_with_grace)

pass_percentage = np.mean(marks_with_grace >= 50) * 100
print("T9b Pass %:", pass_percentage)

grace_only_pass = np.where((marks < 50) & (marks_with_grace >= 50))[0]
print("T9c Students passed only due to grace (indices):", grace_only_pass)

mean_m = np.mean(marks_with_grace)
std_m = np.std(marks_with_grace)
z_scores = (marks_with_grace - mean_m) / std_m
print("T9d Z-scores:", z_scores)
````
