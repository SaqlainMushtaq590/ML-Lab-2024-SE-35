
## Task 06-Tuple vs List
**Question:** What is the main difference between a tuple and a list?

| Feature | **List** `[ ]` | **Tuple** `( )` |
|---|---|---|
| Mutability | Mutable (can be changed) | Immutable (cannot be changed) |
| Syntax | `[1, 2, 3]` | `(1, 2, 3)` |
| Methods | Many (`append`, `remove`, …) | Few (`count`, `index`) |
| Speed / memory | Slightly slower, more memory | Slightly faster, less memory |

**✔ Main difference:** a list is **mutable**, while a tuple is **immutable**.

```python
my_list = [1, 2, 3]
my_list[0] = 99
print("List after change :", my_list)

my_tuple = (1, 2, 3)
try:
    my_tuple[0] = 99
except TypeError as e:
    print("Tuple error       :", e)
```


## Task 07-Grab the Domain
**Question:** Write a function that grabs the domain from `"user@domain.com"`.

**Approach:** Split on `@` and take the last part.

```python
def domainGet(email):
    return email.split('@')[-1]

print(domainGet("user@domain.com"))
```


## Task 08-Find the Word "dog"
**Question:** Write a function that returns `True` if the word "dog" is in a string (case-insensitive).

**Approach:** Convert to lowercase first so the check is case-insensitive.

```python
def findDog(text):
    return "dog" in text.lower()

print(findDog("Is there a dog here?"))
print(findDog("DOG days are here"))
print(findDog("No pets here"))
```


## Task 09-Count the Word "dog"
**Question:** Write a function that counts how many times "dog" occurs in a string (case-insensitive).

**Approach:** Lowercase the string, then use `.count()`.

```python
def countDog(text):
    return text.lower().count("dog")

print(countDog("This dog runs faster than the other dog"))
print(countDog("Dog, DOG, dOg!"))
```


## Task 10-Caught Speeding
**Question:** `caught_speeding(speed, is_birthday)`: ≤ 60 → "No Ticket", 61–80 → "Small Ticket", ≥ 81 → "Big Ticket". On a birthday, allow 5 mph more.

**Approach:** On a birthday, subtract 5 from the speed before checking the limits.

```python
def caught_speeding(speed, is_birthday):
    if is_birthday:
        speed -= 5          # 5 mph extra allowed on birthday

    if speed <= 60:
        return "No Ticket"
    elif speed <= 80:
        return "Small Ticket"
    else:
        return "Big Ticket"

print(caught_speeding(60, False))   # No Ticket
print(caught_speeding(65, True))    # No Ticket
print(caught_speeding(65, False))   # Small Ticket
print(caught_speeding(85, True))    # Small Ticket
print(caught_speeding(85, False))   # Big Ticket
```
